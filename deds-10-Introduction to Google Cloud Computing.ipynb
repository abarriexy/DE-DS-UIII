## 1. Context and Problem Statement
Academic research in data science frequently requires automated collection of textual metadata for literature reviews or text mining. Manual compilation is inefficient and prone to error. This module provides a programmatic solution to ingest book metadata—such as titles, authors, and descriptions—by querying the Google Books API.

### Step-by-Step Configuration:

1. Navigate to the GCP Console https://console.cloud.google.com/ (*Please ensure that you are logged in using the same account that was used during Week 04*).
2. Select previous project (from week 04)
3. Access **APIs & Services > Library**, search for **Books API**, and select **Enable**.
4. Navigate to **APIs & Services > Credentials** and update the enabled API from this API key.
5. Copy the **API Key**.

```python
import requests
import pandas as pd

def fetch_books(query, api_key, max_results=5):
    # Queries the Google Books API and extracts metadata.
    base_url = "https://www.googleapis.com/books/v1/volumes"
    params = {"q": query,"key": api_key, "maxResults": max_results}
    # Execute HTTP GET request
    response = requests.get(base_url, params=params)
    if response.status_code != 200:
        raise RuntimeError(f"API request failed with status code {response.status_code}")
    data = response.json()
    books_list = []
    # Parse JSON payload
    for item in data.get("items", []):
        volume_info = item.get("volumeInfo", {})
        books_list.append({
            "Title": volume_info.get("title", "N/A"),
            "Authors": ", ".join(volume_info.get("authors", ["Anonymous"])),
            "Published Date": volume_info.get("publishedDate", "N/A"),
            "Description": volume_info.get("description", "No description available.")[:100] + "..."})
    return pd.DataFrame(books_list)

# Execution block
try:
    df_book = fetch_books(query=SEARCH_QUERY, api_key=API_KEY)
    print("Metadata extraction successful:")
except Exception as e:
    print(f"Execution error: {e}")
```

## Google Cloud Platform (GCP) Configuration

The Google Sheets API orchestrates secure machine-to-machine interaction using a structural `service_account.json` credential file.

1. Ensure the user-facing Google Form has its response destination directed to a dedicated **Google Sheet**.
2. Navigate to the [GCP Console](https://console.cloud.google.com/) and enable the **Google Sheets API**.
3. Create a **Service Account** via the IAM dashboard, generate a unique **JSON Private Key**, and name the payload file `service_account.json`.
4. Copy the unique email string of the Service Account, select the **Share** menu on your Google Sheet, and grant **Editor** permissions to this machine address to authorize both reading and writing sequences.

```python
from googleapiclient.discovery import build
from google.oauth2.service_account import Credentials
import pandas as pd

def read_google_sheet(spreadsheet_id, sheet_range, key_path):
    #Extracts data from a Google Sheet and returns a Pandas DataFrame.
    scopes = ['https://www.googleapis.com/auth/spreadsheets.readonly']
    credentials = Credentials.from_service_account_file(key_path, scopes=scopes)
    service = build('sheets', 'v4', credentials=credentials)
    
    # Execute API call to read data
    response = service.spreadsheets().values().get(
        spreadsheetId=spreadsheet_id,range=sheet_range).execute()
    matrix = response.get('values', [])
    if not matrix:
        return pd.DataFrame()
    # Parse the first row as columns
    return pd.DataFrame(matrix[1:], columns=matrix[0])

# Configuration
SPREADSHEET_ID = "YOUR_SPREADSHEET_ID"
READ_RANGE = "Sheet1"
CREDENTIALS_FILE = "service_account.json"

# Execution
df_received = read_google_sheet(SPREADSHEET_ID, READ_RANGE, CREDENTIALS_FILE)
print("Data successfully retrieved from cloud storage:")
```
