# <center><font color="blue"> Cloud Computing? </font></center>

> “A model for enabling convenient **on-demand network access** to a **shared** pool of  **configurable computing resources** that can be rapidly provisioned and released with **minimal** management effort or **service provider interaction** ” [US National Institute of Standards and Technology (NIST)]

Cloud computing is a modern approach to delivering computing services through the internet. Instead of purchasing and maintaining physical computer hardware and software independently, users can access computing resources remotely whenever needed. These resources may include servers, storage systems, databases, networking tools, software applications, and artificial intelligence services.

Cloud computing has become one of the foundational technologies of the digital era because it allows organizations and individuals to use powerful computing capabilities without owning expensive infrastructure.

# <center><font color="blue"> When Should Cloud Computing Be Used? </font></center>

* When computational, memory, and/or storage requirements are uncertain or highly dynamic (*elastic*).
* When the goal is to minimize infrastructure maintenance and focus primarily on application development and deployment.
* When ensuring system availability, low latency, and robustness for users is essential.
* When rapid application development and deployment are required.
* When there is a desire to reduce the complexity associated with managing network and infrastructure security configurations.

# <center><font color="blue"> When (Public) Cloud Is Less Suitable </font></center>

* No stable and sufficient funding for Cloud operations.
* Stable long-term computer resource requirements.
* Confidential data that regulations prohibit from being uploaded to the cloud.
* Unreliable internet connections for application users.
* Vendor lock-in ... high dependency on providers.
* Limited system customization.
* Of course, hybrid approaches are possible
* For example, combining simple web hosting, local computing power, and/or VPS.

# <center><font color="blue"> Characteristics of Cloud Computing </font></center>

<center><img src="images/deds-09/karakter-cloud.jpg" alt="" height="500" /></center>

* **On-demand self-service**: No human interaction required (e.g., customer service)
* **Broad network access**: Accessible from various heterogeneous client/user platforms (e.g., mobile phones, tablets, laptops, IoT devices, and workstations).  
* **Resource pooling**: Cloud provider resources are pooled and shared among multiple users.
* **Rapid elasticity**: Resources can be scaled up or down quickly according to demand.
* **Measured service**: Resource usage can be monitored, controlled, and billed automatically.

<center><h1><strong><font color="blue">Theoretical Foundations of Cloud Computing: Virtualization</font></strong></h1></center>

Virtualization is the process of creating virtual versions of computing resources such as servers, storage devices, or operating systems. A single physical computer can be divided into multiple virtual machines (VMs), each functioning as an independent computer.

### Virtualization enables cloud providers to:
- Maximize hardware utilization
- Reduce operational costs
- Provide flexible resource allocation
- Rapidly deploy services

## Elasticity and Scalability

* **Scalability** refers to the ability of a system to handle increasing workloads by adding resources.
* **Elasticity** refers to the automatic adjustment of resources according to demand.
* **Importance**: These characteristics are among the most significant advantages of cloud computing because they allow organizations to efficiently manage changing workloads.

# <center><font color="blue"> Cloud Service Paradigms </font></center>

<center><img src="images/deds-09/cloud-paradigms.png" alt="" width="600" /></center>

## Software as A Service (SaaS)
* Examples: Google Docs, Dropbox, Onedrive, Gdrive, Slack, Skype, Zoom, Microsoft Office 365
* Software is not downloaded and installed; it requires an active internet connection.
* Advantages: Companies are not burdened with installation, management, maintenance, etc.
* Disadvantages: Vendor lock-in, performance, data security.
* Applicability: Suitable for small startups, short-term projects, and temporary-use applications.

## Platform as A Service (PaaS)
* Provides a platform/environment for application development and deployment.
* Developers focus on application development without managing infrastructure.

## Infrastructure as A Service (IaaS)
* Provides virtualized computing infrastructure.
* Users manage operating systems, applications, and configurations independently.

# Comparative Analysis of IaaS, PaaS, and SaaS

| Aspect | IaaS | PaaS | SaaS |
|---|---|---|---|
| Main Purpose | Provide infrastructure | Provide development platform | Provide ready-to-use software |
| User Control | Highest | Moderate | Lowest |
| Technical Skill Required | High | Moderate | Low |
| Infrastructure Management | User manages many components | Provider manages infrastructure | Provider manages everything |
| Flexibility | Very flexible | Moderately flexible | Limited flexibility |
| Ease of Use | Complex | Easier | Easiest |
| Typical Users | System administrators, IT engineers | Software developers | General users |
| Example Services | Amazon EC2 | Google App Engine | Google Docs |

# Responsibility Comparison

| Component | IaaS | PaaS | SaaS |
|---|---|---|---|
| Applications | User | User | Provider |
| Data | User | User | Provider/User |
| Runtime | User | Provider | Provider |
| Operating System | User | Provider | Provider |
| Virtualization | Provider | Provider | Provider |
| Servers | Provider | Provider | Provider |
| Networking | Provider | Provider | Provider |
